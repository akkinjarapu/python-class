```python
word = "abracadabra"
```

```python
def frequency(word):
    freq = {}
    for letter in word:
        if letter in freq:
            freq[letter] += 1
        else:
            freq[letter] = 1
    return freq
```

```python
frequency(word)
```

```python
def frequency(word):
    freq = {}
    for letter in word:
        freq[letter] = freq.get(letter, 0) + 1
    return freq
```

```python
frequency(word)
```

```python
sorted_freq = sorted(frequency(word))
sorted_freq
```

```python
sorted_freq = sorted(frequency(word).items(), key=lambda x: x[0])
sorted_freq
```

```python
sorted_freq = sorted(frequency(word).items(), key=lambda x: x[0], reverse=True)
sorted_freq
```

```python
sorted_freq = sorted(frequency(word).items(), key=lambda x: x[1])
sorted_freq
```

```python
sorted_freq = sorted(frequency(word).items(), key=lambda x: x[1], reverse=True)
sorted_freq
```

```python
def factorial(n):
    if n <= 0:
        return 1
    else:
        return n * factorial(n - 1)
```

```python
factorial(5), factorial(0), factorial(1)
```

```python
def factorial(n):
    for i in range(1, n):
        n *= i
    return 1 if n == 0 else n
```

```python
factorial(5), factorial(0), factorial(1)
```

```python
def factorial(n):
    for i in range(n-1, 1, -1):
        n *= i
    return 1 if n == 0 else n
```

```python
factorial(5), factorial(0), factorial(1)
```

```python
import math
factorial = math.factorial
factorial(5), factorial(0), factorial(1)
```

```python
import functools
def factorial(n):
    return functools.reduce(lambda x, y: x * y, range(1, n + 1), 1)
```

```python
factorial(5), factorial(0), factorial(1)
```

```python
def gcd(a, b):
    while b:
        a, b = b, a % b
    return a
```

```python
gcd(48, 18), gcd(101, 10), gcd(0, 5), gcd(5, 0), gcd(0, 0)
```

```python
def gcd(a, b):
    return gcd(b, a % b) if b else a
```

```python
gcd(48, 18), gcd(101, 10), gcd(0, 5), gcd(5, 0), gcd(0, 0)
```
