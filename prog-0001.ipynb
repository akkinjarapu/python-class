```python
def grade(score):
    if score >= 90:
        return 'S'
    elif score >= 80:
        return 'A'
    elif score >= 70:
        return 'B'
    elif score >= 60:
        return 'C'
    elif score >= 50:
        return 'D'
    elif score >= 40:
        return 'E'
    else:
        return 'F'
```

```python
grade(85)
```

```python
grades = [(score, grade(score)) for score in [95, 85, 75, 65, 55, 45, 35]]
grades
```
