```python
def greeting(name):
    name = f"Hello, {name}!"
    return name
```

```python
name = 'AITAM'
greet = greeting(name)
name, greet
```

```python
def plus10(value):
    value += 10
    return value
```

```python
num = 50
result = plus10(num)
num, result
```

```python
def add_element(lst):
    lst.append(len(lst) + 1)
    return lst
```

```python
nums = [1, 2, 3]
result = add_element(nums)
nums, result
```

```python
def mysum(*args):
    total = 0
    for num in args:
        total += num
    return total
```

```python
mysum(1,2,3,4,5)
```

```python
def mysum(*args):
    return sum(args)
```

```python
mysum(1,2,3,4,5)
```
